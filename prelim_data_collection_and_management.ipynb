```python
# England

# Step 1: Fetching the HTML content of the webpage
url1 = 'https://www.worldfootball.net/alltime_table/eng-premier-league/'
response = requests.get(url1)

# Checking if the request was successful
if response.status_code == 200:
    soup = BeautifulSoup(response.text, 'html.parser')

    # Step 2: Locating the table in the HTML
    table1 = soup.find('table', {'class': 'standard_tabelle'})

    # Step 3: Extracting Headers
    headers1 = [th.text.strip() for th in table1.find_all('th')]
    print(f"Headers (Columns): {headers1}")
    print(f"Number of Columns in Headers: {len(headers1)}")

    # Step 4: Extracting Table rows
    rows = []
    for tr in table1.find_all('tr')[1:]:  # Skipping header row
        cols = [td.text.strip() for td in tr.find_all('td')]
        if cols:  # Skipping empty rows
            rows.append(cols)

    print(f"First Row: {rows[0]}")
    print(f"Number of Columns in First Row: {len(rows[0])}")

    # Step 5: Ensuring Column Alignment
    if len(headers1) < len(rows[0]):
        # Adding placeholder headers for extra columns
        headers1 += [f"Extra Column {i+1}" for i in range(len(rows[0]) - len(headers1))]
    elif len(headers1) > len(rows[0]):
        print("Warning: Headers have more columns than data rows.")
else:
    print(f"Failed to retrieve the page. Status code: {response.status_code}")

df_eng = pd.DataFrame(rows, columns = headers1)
```

```python
dataframes = [df_eng, df_ita, df_esp, df_ger, df_por, df_fra]

for i in dataframes:
  # Checking if 'Team' column exists before dropping
  if 'Team' in i.columns:
      i.drop(columns = ["Team"], inplace = True)
  # Assigning Column names as a flat list, not a list of lists
  i.columns = ["All-Time Domestic Position", "Team", "Matches", "Win", "Draw", "Loss", "GS:GC", "GD", "Total Points"]

df_ita.head(6) # Check
```

```python
for i in dataframes:
  i[["Matches", "Win", "Draw", "Loss", "GD", "Total Points"]] = i[["Matches", "Win", "Draw", "Loss", "GD", "Total Points"]].astype(int)

df_eng.dtypes # Check
```

```python
for index, df in enumerate(dataframes):
    # 1. Splitting 'GS:GC' first to create new Columns
    # Accessing "GS:GC" will now work correctly with a flat column index
    df[["Goals Scored", "Goals Conceded"]] = df["GS:GC"].str.split(":", expand=True)
    df[["Goals Scored", "Goals Conceded"]] = df[["Goals Scored", "Goals Conceded"]].astype(int)

    # 2. Then drop 'GS:GC'
    df.drop(columns=["GS:GC"], inplace=True)
```

```python
# Rearranging the Columns via a "for" Loop was giving Problems, so I just Manually did it

df_eng[["All-Time Domestic Position", "Team", "Matches", "Win", "Draw", "Loss", "Goals Scored", "Goals Conceded", "GD", "Total Points"]]
df_ita[["All-Time Domestic Position", "Team", "Matches", "Win", "Draw", "Loss", "Goals Scored", "Goals Conceded", "GD", "Total Points"]]
df_esp[["All-Time Domestic Position", "Team", "Matches", "Win", "Draw", "Loss", "Goals Scored", "Goals Conceded", "GD", "Total Points"]]
df_ger[["All-Time Domestic Position", "Team", "Matches", "Win", "Draw", "Loss", "Goals Scored", "Goals Conceded", "GD", "Total Points"]]
df_por[["All-Time Domestic Position", "Team", "Matches", "Win", "Draw", "Loss", "Goals Scored", "Goals Conceded", "GD", "Total Points"]]
df_fra[["All-Time Domestic Position", "Team", "Matches", "Win", "Draw", "Loss", "Goals Scored", "Goals Conceded", "GD", "Total Points"]]
```

```python
for index, i in enumerate(dataframes):
  i = i.assign(
    PPG = (i["Total Points"]/i["Matches"]).round(3),
    WPG = (i["Win"]/i["Matches"]).round(3),
    DPG = (i["Draw"]/i["Matches"]).round(3),
    LPG = (i["Loss"]/i["Matches"]).round(3),
    GSPG = (i["Goals Scored"]/i["Matches"]).round(3),
    GCPG = (i["Goals Conceded"]/i["Matches"]).round(3))
  dataframes[index] = i

# Accessing and Displaying modified Dataframes
df_eng = dataframes[0]
df_ita = dataframes[1]
df_esp = dataframes[2]
df_ger = dataframes[3]
df_por = dataframes[4]
df_fra = dataframes[5]

df_ita.head(6) # Check
```

```python
# Flattening the Multi-Index Column names

for i in dataframes:
  i.columns = [' '.join(col).strip() if isinstance(col, tuple) else col for col in i.columns]
```

```python
# Refining some Columns

for i in dataframes:
  i["Country GSPG"] = i["GSPG"].mean().round(3)
  i["Country GCPG"] = i["GCPG"].mean().round(3)
  i["ΔGSPG From Country Avg."] = (i["GSPG"] - i["Country GSPG"]).round(3)
  i["ΔGCPG From Country Avg."] = (i["GCPG"] - i["Country GCPG"]).round(3)

df_ita.head(5) # Check
```

```python
# Important Column to Add before the Merge

df_eng["Country"] = "England"
df_ita["Country"] = "Italy"
df_esp["Country"] = "Spain"
df_ger["Country"] = "Germany"
df_por["Country"] = "Portugal"
df_fra["Country"] = "France"
```

# **Machine Learning** & Predictive Modeling

### Feature Engineering & Preprocessing

```python
# Will add a Column for "Maximum Points" Possible for each Team
# This is an Estimate of the Total Points a Team would have Accumulated if it played the Maximum Possible Games with its current Average Quality

df_eng["Expected Max Points"] = (df_eng["PPG"] * df_eng['Matches'].max()).astype(int)
df_ita["Expected Max Points"] = (df_ita["PPG"] * df_ita['Matches'].max()).astype(int)
df_esp["Expected Max Points"] = (df_esp["PPG"] * df_esp['Matches'].max()).astype(int)
df_fra["Expected Max Points"] = (df_fra["PPG"] * df_fra['Matches'].max()).astype(int)
df_ger["Expected Max Points"] = (df_ger["PPG"] * df_ger['Matches'].max()).astype(int)
df_por["Expected Max Points"] = (df_por["PPG"] * df_por['Matches'].max()).astype(int)
```

```python
# Regression Plot to get a Better Depiction of certain Relationships as a Precursor to the Modeling to Come
sns.regplot(x = df_eng['WPG'], y = df_eng['Expected Max Points'])
plt.xlabel('Wins per Game')
plt.title('Regression Plot of Wins per Game (WPG) to Expected Max Points')
plt.show()
```

## Regression Model

```python
features = ['WPG', 'DPG', 'LPG', 'ΔGSPG From Country Avg.', 'ΔGCPG From Country Avg.']

X = df_eng[features]
y = df_eng['PPG']

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 42)
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()

lr_model = lr.fit(X_train, y_train)
y_pred_lr = lr.predict(X_test)
```

```python
# Residuals
diff_lr = y_test - y_pred_lr
```

```python
# Create Series for predicted values and the difference with the same index as y_test
y_pred_lr_series = pd.Series(y_pred_lr, index = y_test.index)
diff_lr_series = pd.Series(diff_lr, index = y_test.index)

# Concatenate the Series, now aligned by index
residuals_dt = pd.concat([y_test, y_pred_lr_series, diff_lr_series], axis = 1)
residuals_dt.columns = ["Actual Result", "Predicted Result", "Difference"]
#residuals_dt = residuals_dt.round(2) # Note: Corrected the variable name here
residuals_dt.head(8)
```

```python
from sklearn.metrics import mean_squared_error as mse, mean_absolute_error as mae, root_mean_squared_error as rmse, r2_score as r2, accuracy_score as acc

print("The Mean Squared Error is: {}".format(round(mse(y_test, y_pred_lr), 2)))
print("The Mean Absolute Error is: {}".format(round(mae(y_test, y_pred_lr), 2)))
print("The Root Mean Squared Error is: {}".format(round(rmse(y_test, y_pred_lr), 2)))
print("The R2 Score is: {}".format(round(r2(y_test, y_pred_lr)*100, 2)))
print("Linear Regression Model score: {}".format(round(lr_model.score(X_train, y_train)*100, 2)))
```

```python
# print(lr_model.coef_)
print(lr_model.intercept_.round(3))

pd.DataFrame(lr_model.coef_.round(3), X.columns, columns = ['Coefficient'])
```
